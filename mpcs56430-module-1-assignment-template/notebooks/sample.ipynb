# mpcs56430-assignment-1: Kinetic Energy Analysis
This notebook demonstrates package imports and basic scientific plotting.

The first code cell auto-detects the project root and adds `src/` to `sys.path`.


```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "pyproject.toml").exists() and (candidate / "src").exists():
            return candidate
    raise RuntimeError(
        "Could not locate project root containing pyproject.toml and src/."
    )


project_root = find_project_root(Path.cwd())
src_dir = project_root / "src"
if str(src_dir) not in sys.path:
    sys.path.insert(0, str(src_dir))

from mpcs56430_assignment_1.physics import compute_kinetic_energy

print(f"Environment ready. Using src path: {src_dir}")
```

```python
mass = 10.0  # kg
velocities = np.linspace(0, 100, 50)
energies = [compute_kinetic_energy(mass, v) for v in velocities]

plt.figure(figsize=(8, 5))
plt.plot(velocities, energies, color="tab:blue", lw=2)
plt.title(f"Kinetic Energy vs Velocity (Mass={mass} kg)")
plt.xlabel("Velocity (m/s)")
plt.ylabel("Energy (Joules)")
plt.grid(True, alpha=0.3)
plt.show()
```
